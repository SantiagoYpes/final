# Desarrollo del modelo de linea Base

Para poder comparar la efectividad de nuestros modelos desarrollaremos la línea base que nos dará la métrica a vencer de los futuros ejercicios

## Librerias y Configuración

```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, classification_report
)
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.float_format", "{:.4f}".format)
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../data/merged_df.csv")


print(f"DF cargado: {df.shape}")
print(f"Columnas: {len(df.columns)}")
```

En este caso vamos a configurar que todo el mundo no reingresa que es la clase mayoritaria

```python
# Definir el target
y_true = df["readmission_30d"]

print("\nDistribución del target:")
print(y_true.value_counts())
print(y_true.value_counts(normalize=True).round(4))

y_pred_baseline = np.zeros(len(y_true), dtype=int)

acc   = accuracy_score(y_true, y_pred_baseline)
prec  = precision_score(y_true, y_pred_baseline, zero_division=0)
rec   = recall_score(y_true, y_pred_baseline, zero_division=0)
f1    = f1_score(y_true, y_pred_baseline, zero_division=0)
```

```python
print(f"Accuracy : {acc:.4f}  ({acc*100:.1f}%)")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f} ")
print(f"F1-score : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_true, y_pred_baseline, 
                          target_names=["No reingresa (0)", "Reingresa (1)"],
                          zero_division=0))
```

En este caso evaluamos la posibilidad de que un paciente reingrese en los proximos 30 dias al servicio de hospitalización, teniendo una precisión en la linea base de 0.67
