# Desarrollo del modelo de linea Base

Para poder comparar la efectividad de nuestros modelos desarrollaremos la línea base que nos dará la métrica a vencer de los futuros ejercicios

## Librerias y Configuración

```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, classification_report
)
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.float_format", "{:.4f}".format)
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../data/merged_df.csv")


print(f"DF cargado: {df.shape}")
print(f"Columnas: {len(df.columns)}")
```

En este caso vamos a configurar que todo el mundo no muere

```python
# Definir el target
y_true = df["in_hospital_mortality"]

print("\nDistribución del target:")
print(y_true.value_counts())
print(y_true.value_counts(normalize=True).round(4))

y_pred_baseline = np.zeros(len(y_true), dtype=int)

acc   = accuracy_score(y_true, y_pred_baseline)
prec  = precision_score(y_true, y_pred_baseline, zero_division=0)
rec   = recall_score(y_true, y_pred_baseline, zero_division=0)
f1    = f1_score(y_true, y_pred_baseline, zero_division=0)
```

```python
print(f"Accuracy : {acc:.4f}  ({acc*100:.1f}%)")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f} ")
print(f"F1-score : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_true, y_pred_baseline, 
                          target_names=["No muere (0)", "Muere (1)"],
                          zero_division=0))
```

En este caso evaluamos que aunque la clase mayoritaria es bastante amplia, el valor en este ejercicio es cambiar el recall de la linea base que nunca detecta cuando un paciente muere, que para terminos de un hospital sería información valiosa para encontrar pacientes que ingresan a hospitalización con altas probabilidades de fallecer.

`Valor a vencer` : Accuracy - 0.90
